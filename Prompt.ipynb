```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI

os.environ["GOOGLE_API_KEY"] = "SECRET
def pick_free_gemini_model():
    """Pick the best free Gemini model (flash or flash-lite) from available list."""
    client = genai.Client()
    available_models = [m.name for m in client.models.list()]

    # Prefer the latest flash model
    for preferred in [
        "models/gemini-2.5-flash",
        "models/gemini-2.5-flash-lite",
        "models/gemini-flash-latest",
        "models/gemini-flash-lite-latest",
    ]:
        if preferred in available_models:
            return preferred

    raise ValueError("No free Gemini flash model found in your account.")

# Auto-select model
chosen_model = pick_free_gemini_model()
print(f"✅ Using model: {chosen_model}")

# Initialize in LangChain
model = ChatGoogleGenerativeAI(
    model=chosen_model,
    temperature=1.0
)

# Test call
result = model.invoke("Write a 5 line poem on cricket")
print(result.content)

```
